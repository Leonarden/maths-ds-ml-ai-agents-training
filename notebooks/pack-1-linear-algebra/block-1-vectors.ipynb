#### Python tasks

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([-1, 0, 4])
c = np.array([2, 2, 2])

print((a + b) + c, a + (b + c))  # associativity
print(a + b, b + a)              # commutativity
print(0 * a)                     # zero scalar
print(-1 * a)                    # additive inverse
```
