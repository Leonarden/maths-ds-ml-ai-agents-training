#### Exercises

Independence test — Determine if sets of 2, 3, and 4 vectors are independent.

Dependence proof — Show that if one vector is a linear combination of others, the system is dependent.

Zero vector case — Prove that any system containing 0 is dependent.



#### Python tasks

```python
import numpy as np

A = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [1, 0, 1]
])

rank = np.linalg.matrix_rank(A)
print("Rank:", rank)
print("Independent?" if rank == A.shape[1] else "Dependent")
```
